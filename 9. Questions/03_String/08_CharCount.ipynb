```python
# Write a Python program to count the number of characters (character frequency) in a string.
# Sample String : google'
# Expected Result : {'g': 2, 'o': 2, 'l': 1, 'e': 1}
```

```python
def char_count(str1):
    dict={}
    for char in str1:
        key=dict.keys()
        if char in key:
            dict[char]+=1
        else:
            dict[char]=1
    return dict
print(char_count("hello jobee dobee"))
```
