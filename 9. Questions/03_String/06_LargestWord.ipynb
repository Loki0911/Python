```python
# Write a Python function that takes a list of words and returns the length of the longest one
```

```python
def find_largest_word(words):
    word_len=[]
    for word in words:
        word_len.append((len(word),word))
    word_len.sort()
    return word_len[-1][1]
print(find_largest_word(["PHP","python","java"]))
```
