```python
# Write a Python program to remove the nth index character from a nonempty string
```

```python
def remove_index(str,n):
    first_str=str[:n]
    last_str=str[n+1:]
    return first_str+last_str
print(remove_index("python",0))
print(remove_index("python",1))
print(remove_index("python",2))
print(remove_index("python",3))
print(remove_index("python",4))
print(remove_index("python",5))
```
