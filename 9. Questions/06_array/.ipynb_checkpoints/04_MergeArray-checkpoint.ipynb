```python
#Given a collection of intervals which are already sorted by start number, merge all overlapping intervals.
#For example,
#Given [[1,3],[2,6],[5,10],[11,16],[15,18],[19,22]],
#return [[1, 10], [11, 18], [19, 22]]
```

```python
org_interval=[[1,3],[2,6],[5,10],[11,16],[15,18],[19,22]]
i=0
while i < len(org_interval) - 1:
    if org_interval[i+1][0] < org_interval[i][1]:
        org_interval[i][1] = org_interval[i+1][1]
        del org_interval[i+1]
        i = i - 1
    i = i + 1
print(org_interval)
```
