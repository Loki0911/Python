```python
#  Write a Python program to solve the Fibonacci sequence using recursion.
```

```python
def fib(n):
    if n==0:
        return 0
    elif n==1:
        return 1

    else:
        return fib(n-1)+fib(n-2)
n=10
print(f"Fibonacci sequence up to {n}")
for i in range(n):
    print(fib(i),end=" ")
```
