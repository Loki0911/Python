```python
# Write a Python program to find common items from two lists.
# input
# color1 = "Red", "Green", "Orange", "White"
# color2 = "Black", "Green", "White", "Pink"
# output
# {'Green', 'White'}
```

```python
color1 = "Red", "Green", "Orange", "White"
color2 = "Black", "Green", "White", "Pink"
common=list(set(color1) & set(color2))
print(common)
```
