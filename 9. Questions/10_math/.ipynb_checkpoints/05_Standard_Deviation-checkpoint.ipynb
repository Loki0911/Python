```python
# Write a Python program to calculate the standard deviation of the following data.
# Input
# Sample Data:  [4, 2, 5, 8, 6]                                                                                 
# Output
# Standard Deviation :  2.23606797749979
```

```python
import math
Sample_data=[4, 2, 5, 8, 6] 
n=len(Sample_data)
def calculate_mean(data):
    
    mean=float(sum(Sample_data)/float(n))
    return mean
def calculate_sd(data):
    x=0
    mean=calculate_mean(data)
    for el in data:
        x+=(float(el)-mean)**2
    sd=math.sqrt(x/float(n-1))
    return sd
print("Mean = ",calculate_mean(Sample_data))
print("Standard Deviation =",calculate_sd(Sample_data))
```
