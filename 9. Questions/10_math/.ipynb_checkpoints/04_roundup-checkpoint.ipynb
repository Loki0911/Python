```python
# Write a Python function to round up a number to specified digits.
```

```python
import math

def roundup(a,digit=0):
    n=10**(-digit)
    return round(math.ceil(a/n)*n,digit)

x=float(input("Enter the number"))
print("Orginal Number : ",x)
print(roundup(x,0))
print(roundup(x,1))
print(roundup(x,2))
print(roundup(x,3))
```
