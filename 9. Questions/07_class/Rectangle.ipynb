```python
class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def compute_perimeter(self):
        print("Perimeter = ",2*(self.length+self.breadth))
    def compute_area(self):
        print("Area =",self.length*self.breadth)
length,breadth=float(input("Enter the length")),float(input("Enter the breadth"))
rect=Rectangle(length,breadth)
rect.compute_perimeter()
rect.compute_area()
```
