```python
# Write a Python program to find validity of a string of parentheses, '(', ')', '{', '}', '[' and ']. 
# These brackets must be close in the correct order, 
```

```python
class pySolution:
    def is_valid_parenthese(self,str1):
        stack=[]
        pychar={'(':')','{':'}','[':']'}
        for parenthese in str1:
            if parenthese in pychar:
                stack.append(parenthese)
            elif len(stack) == 0 or pychar[stack.pop()]!=parenthese:
                return False
        return len(stack)==0
print(pySolution().is_valid_parenthese("(){}[]"))
print(pySolution().is_valid_parenthese("(}{)[]"))
print(pySolution().is_valid_parenthese("(){]"))
```
