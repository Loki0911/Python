```python
# Write a Python program to find all five characters long word in a string.
# Input
# 'The quick brown fox jumps over the lazy dog.'
# Output
# ['quick', 'brown', 'jumps']
```

```python
import re
text="A quick brown fox jumps over a lazy dog"
print(re.findall(r"\b\w{5}\b",text))
```

```python
#\b → word boundary (ensures whole words are matched).

#\w{5} → matches exactly 5 word characters (letters, digits, or underscore).

#\b → ensures the match ends at a word boundary.
```
