```python
# Write a Python program to sort (ascending and descending) a dictionary by value.
# Original dictionary :  {0: 0, 1: 2, 2: 1, 3: 4, 4: 3}                                                         
# Dictionary in ascending order by value :  [(0, 0), (1, 2), (2, 1), (3, 4), (4, 3)]                            
# Dictionary in descending order by value :  [(4, 3), (3, 4), (2, 1), (1, 2), (0, 0)]
```

```python
import operator

d={0: 0, 1: 2, 2: 1, 3: 4, 4: 3} 
print(f"Orginal Dictionary : {d}")
ascending_sorted=sorted(d.items(),key=operator.itemgetter(0))
print(f" Dictionary sorted in ascending order : {ascending_sorted}")
descending_sorted=sorted(d.items(), key=operator.itemgetter(0), reverse=True)
print(f" Disctionary in descending order : {descending_sorted}")
```
