```python
# Write a Python program that accepts a string and calculate the number of digits and letters
# Sample Data : "Python 3.2"
# Expected Output :
# Letters 6 
# Digits 2

```

```python
s=input("Enter a string")
d=l=0
for c in s:
    if c.isdigit():
        d+=1
    elif c.isalpha():
        l+=1
    else:
        pass

print(f"letters={l}")
print(f"digits={d}")
```
