```python
len1=int(input("Enter the length of first side"))
len2=int(input("Enter the length of second side"))
len3=int(input("Enter the length of third side"))



def triangle_check(l1,l2,l3):
    if (l1>l2+l3) or (l2>l1+l3) or (l3>l1+l2):
        print("No, the length wont form a triangle")
    elif (l1==l2+l3) or (l2==l1+l3) or (l3==l1+l2):
        print('yes, it can form a degenerated triangle')
    else:
        print('Yes, a triangle can be formed out of it')
triangle_check(len1,len2,len3)
```
