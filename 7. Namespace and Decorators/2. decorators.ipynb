### Decorators:
- Decorators are functions that takes another function as an argument and adds some functionality(decorations) to it and returns it.
- This happens because function are first class citizens .

````python
# functions as first class citizens
def modify1(func,num):  
    return func(num)

def square(num):
    return num*num

a = modify1(square,5)

a()
````

````python
# simple example

def my_decorator(func):  # -> takes function as argument
    def wrapper():  # -> adds some functionality.
        print('**************************************')
        func()
        print('**************************************')
    return wrapper   # -> returns the function

@my_decorator   # -> clear syntax
def hell0():
    print('Hello World')

@my_decorator   # clear syntax
def display():
    print("Hello lokesh")

hell0()
display()
````

````python
# decorator that prints the execution time of the function

import time

def timer(func):

    # wrapper function that adds timing functionality
    def wrapper(*args):

        # record the start time before function execution
        start = time.time()

        # call the original function
        func(*args)

        # calculate and print the execution time
        print("Time Taken by", func.__name__, time.time() - start, 'seconds')

    # return the wrapper function
    return wrapper

# hello function is decorated with timer
@timer
def hello():
    print("Hello world")
    time.sleep(2)  # simulate a task that takes 2 seconds

# display function is decorated with timer
@timer
def display():
    print('Displaying smtg')
    time.sleep(4)  # simulate a task that takes 4 seconds

# square function is decorated with timer
@timer
def square(num):
    time.sleep(1)  # simulate a task that takes 1 second
    print(num**2)

# execute the decorated functions
hello()
square(5)
display()
````

````python
# decorator that checks whether the argument passed to a function
# is of the expected data type

def sanity_check(data_type):

    # receives the function to be decorated
    def outer_wrapper(func):

        # wrapper function that validates the input
        def inner_wrapper(*args):

            # check if the type of the argument matches the expected type
            if type(*args) == data_type:

                # call the original function if validation passes
                func(*args)

            else:
                # raise an error if the data type is incorrect
                raise TypeError("Wrong datatype")

        return inner_wrapper

    return outer_wrapper

# square function should only accept integers
@sanity_check(int)
def square(num):
    print(num**2)

# valid input (integer)
square(4)

# greet function should only accept strings
@sanity_check(str)
def greet(name):
    print("Hello", name)

# valid input (string)
greet('lokesh')
````

### `Q1:` Write a Python program to make a chain of function decorators (bold, italic, underline etc.) on a given function which prints "hello world"

```
def hello():
    return "hello world"
```

```
bold - wrap string with <b> tag. <b>Str</b>
italic - wrap string with <i> tag. <i>Str</i>
underline- wrap string with <u> tag. <u>Str</u>
```

````python
# Decorator to wrap output string with <b>...</b> tags
def make_bold(func):
    # Wrapper function adds bold formatting
    def wrapper():
        return "<b>" + func() + "</b>"
    return wrapper


# Decorator to wrap output string with <i>...</i> tags
def make_italic(func):
    # Wrapper function adds italic formatting
    def wrapper():
        return "<i>" + func() + "</i>"
    return wrapper


# Decorator to wrap output string with <u>...</u> tags
def make_underline(func):
    # Wrapper function adds underline formatting
    def wrapper():
        return "<u>" + func() + "</u>"
    return wrapper


# Apply decorators in a chain (bottom to top)
# 1. underline
# 2. italic
# 3. bold
@make_bold
@make_italic
@make_underline
def hello():
    # Original function returns "hello world"
    return "hello world"


# Print decorated output
print(hello())
````

### `Q2:` Write a decorator called `printer` which causes any decorated function to print their return values. If the return value of a given function is `None`, printer should do nothing.



````python
from functools import wraps

# Decorator that prints the return value of the decorated function
# If the return value is None, nothing is printed
def printer(func):

    # Preserve metadata of the original function
    @wraps(func)
    def wrapper(*args, **kwargs):

        # Call the original function and store its return value
        result = func(*args, **kwargs)

        # Print the result only if it is not None
        if result is not None:
            print(result)

    return wrapper


# Apply printer decorator to hello()
@printer
def hello(string):

    # Function returns None
    return None


# Call the decorated function
hello('Hello World')
````

### `Q3:` Make a decorator which calls a given function twice. You can assume the functions don't return anything important, but they may take arguments.
```
#Lets say given function
def hello(string):
    print(string)

#on calling after specified decorator is inplaced
hello('hello')
```

Output:
```
hello
hello
```

````python
from functools import wraps

# Decorator that calls the decorated function twice
def call_twice(func):

    # Preserve metadata of the original function
    @wraps(func)
    def wrapper(*args, **kwargs):

        # Call the original function for the first time
        func(*args, **kwargs)

        # Call the original function for the second time
        func(*args, **kwargs)

    return wrapper


# Apply call_twice decorator to hello()
@call_twice
def hello(string):

    # Print the given string
    print(string)


# Call the decorated function
hello('hello')
````

### `Q4:` Write a decorator which doubles the return value of any function. And test that decoratos is working correctly or not using `asert`.

```
add(2,3) -> result in 10. Without decorator it should be 5.
```

````python
# Decorator that doubles the return value of the decorated function
def double(func):

    # Wrapper function accepts any number of positional and keyword arguments
    def wrapper(*args, **kwargs):

        # Call the original function and store its return value
        return_value = func(*args, **kwargs)

        # Return double the original return value
        return return_value * 2

    return wrapper


# Apply double decorator to add_with_decorator()
@double
def add_with_decorator(a, b):

    # Return sum of two numbers
    return a + b


# Function without decorator
def add(a, b):

    # Return sum of two numbers
    return a + b


# Display results
print(add_with_decorator(2, 3))  # 10
print(add(2, 3))                 # 5

# Verify that decorator doubles the return value correctly
assert add_with_decorator(2, 3) == 2 * add(2, 3), "Decorator is not working correctly"

print("Decorator is working correctly")
````
