Namespace are dictionaries of identifiers as keys and their object as values.
There are 4 types of Namespaces.
1. Builtin Namespace
2. Global Namespace
3. Enclosing Namespace
4. Local Namespace

Scope is the area where a variable can be used or defined.
A interpreter searches for name in local scope, then global scope, then enclosing scope and so on until it finds an identifier.
if not found throws an NameNotFoundError exception.

### Enclosing scopes


````python
def outer():    # - > this is the inclosing scope
    a = 3
    def inner():   # - > this is the local scope

      print(a)          # local scope searched first for the name a , not found then searches in enclosing scopes

    inner()

outer()
a = 5
print('main function')
````

````python
#nonlocal keyword
def outer():    # - > this is the inclosing scope
    a = 3
    def inner():   # - > this is the local scope
        nonlocal a
        a+= 1
        print(a)          # local scope searched first for the name a , not found then searches in enclosing scopes

    inner()
    print(a)
outer()
a = 5
print('main function')
````

### `Q1:` Write `Person` Class as given below and then display it's namespace.

```
Class Name - Person

Attributes:
name - public
state - public
city - private
age - private

Methods:
address - public
It give address of the person as "<name>, <city>, <state>"
```



````python
class Person:

    def __init__(self, name, state):
        self.name = name      # public attribute: name
        self.state = state    # public attribute: state
        self.__city = None    # private attribute: city
        self.__age = None     # private attribute: age

    # getter method for private age attribute
    def get_age(self):
        return self.__age

    # setter method for private age attribute
    def set_age(self, age):
        self.__age = age

    # getter method for private city attribute
    def get_city(self):
        return self.__city

    # setter method for private city attribute
    def set_city(self, city):
        self.__age = city

    # public method that returns address as
    # "<name>, <city>, <state>"
    def address(self):
        return f"{self.name}, {self.__city}, {self.state}"

# display class namespace
for i in Person.__dict__:
    print(i)
````

### `Q2:` Write a program to show namespace of object/instance of above(Person) class.

````python
# Create an object/instance of Person class
p = Person('Lokesh', 'Sudurpaschim')

# Set values for private attributes using setter methods
p.set_age(22)
p.set_city("Dhangadhi")

# Display address of the person
print(p.address())

# Display object/instance namespace
for i in p.__dict__:
    print(i)
````

### `Q3:` Write a recursive program to to calculate `gcd` and print no. of function calls taken to find the solution.

```
gcd(5,10) -> result in 5 as gcd and function call 4
```

````python
counter = 0  # Counts the number of function calls

def gcd(a, b):
    global counter
    counter += 1  # Increment counter on every function call

    # Base case: when second number becomes 0,
    # first number is the GCD
    if b == 0:
        return a
    else:
        # Recursive call using Euclid's algorithm
        return gcd(b, a % b)

# Find and display GCD
print('Result =', gcd(5, 10))

# Display total number of function calls
print('Number of function calls =', counter)
````
