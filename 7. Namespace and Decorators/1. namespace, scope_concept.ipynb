Namespace are dictionaries of identifiers as keys and their object as values.
There are 4 types of Namespaces.
1. Builtin Namespace
2. Global Namespace
3. Enclosing Namespace
4. Local Namespace

Scope is the area where a variable can be used or defined.
A interpreter searches for name in local scope, then global scope, then enclosing scope and so on until it finds an identifier.
if not found throws an NameNotFoundError exception.

```python
#local and global scopes
# exampl 1


# Inn this program ,local scope
# def add(x,y):
#     print(x+y)

#global scope
def add(x,y):
    print(x+y)
a=10
b=20
add(a,b)
```

```python
#example 2
# a is global variable and b is local . b is accessible only inside temp
a=2
def temp():
    b=3
    print(b)

temp()
print(a)
```

```python
# example 3
a=2  # -> this is in global scope
def temp():
    a=3     # -> this is in local scope. so it will be overwritten by the value of a in global scope only inside this function and not outside.
    print(a)

temp()
print(a) 
```

```python
#example 4
a = 2
def temp():
    print(a)   # -> first searches the value in local scope , if not found then it searches in enclosing scope , if not found searches in global scope, there is a defined
temp()
print(a)

```

### Enclosing scopes


```python
def outer():    # - > this is the inclosing scope
    a = 3
    def inner():   # - > this is the local scope

      print(a)          # local scope searched first for the name a , not found then searches in enclosing scopes

    inner()

outer()
a = 5
print('main function')
```

```python
#nonlocal keyword
def outer():    # - > this is the inclosing scope
    a = 3
    def inner():   # - > this is the local scope
        nonlocal a
        a+= 1
        print(a)          # local scope searched first for the name a , not found then searches in enclosing scopes

    inner()
    print(a)
outer()
a = 5
print('main function')
```
