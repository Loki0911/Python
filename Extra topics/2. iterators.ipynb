### `1.  Iteration`

Iterationn is a  general term for taking each item of something one after another.

Anytime you use a loop, explict or implict, to go over a group of items, that is iteration.

````python
L = [1, 2, 3]

# Iteration: the loop visits each element of the list sequentially
for i in L:
    print(i)
````

### `2. Iterator`

Iterator is  an object that allows the  programmer to traverse through a sequence of data without having to store the entire data in memory.

````python
import sys

# Create a list containing numbers from 1 to 99,999
# All values are stored in memory at once
l = [i for i in range(1, 100000)]

# Iteration can be performed on the list
# for i in l:
#     print(i)

# Display memory occupied by the list object
print(sys.getsizeof(l), 'bytes')

# range() is an iterable that generates values on demand
# It does not store all values in memory like a list
x = iter(range(1, 100000))

# Display memory occupied by the iterator object
print(sys.getsizeof(x), 'bytes')
````

### ` 3.  Iterable`

Iterable is an object, which one can iterate over.

It generates an iterator object when passed to iter() method.

````python
L = [1, 2, 3]

# Iterable: an object capable of returning its elements one at a time
type(L)

# Convert the iterable into an iterator
type(iter(L))

# Iterator: an object that keeps track of the current position
# and provides the next element when requested
# iter(L) -> iterator
````

## Points to remember:

- All iterator is also an iterable
- Not all iterables are iterators.

## Trick
- Every iterable has an iter function.
- Every iterator has both iter functio as well as a next function

````python
a = 2

# dir() displays all attributes and methods available for the integer object
dir(a)

# Every iterator implements the __iter__() method
# __iter__() returns the iterator object itself

# Every iterator also implements the __next__() method
# __next__() returns the next item in the sequence

# An object is considered an iterator if it has both
# __iter__() and __next__() methods
````

````python
T = {1: 2,2: 4,3: 5}
dir(T)    # -> is there iter object in the output
````

````python
# List is an iterable, not an iterator
l = [1, 2, 3]

# Display all methods and attributes of the list object
dir(l)

# The output contains __iter__() but not __next__()

# Since __next__() is missing, list is not an iterator

# A list can be converted into an iterator using iter()
# it = iter(l)

# The iterator object will have both __iter__() and __next__()
````

````python
l = [1, 2, 3]

# Convert the iterable list into an iterator
iter_l = iter(l)

# Display all methods and attributes of the iterator object
dir(iter_l)

# iter_l is an iterator because it contains both
# __iter__() and __next__() methods

# __iter__() returns the iterator object itself
# __next__() returns the next element in the sequence
````

### Understanding how loop works

````python
num = [1,2,3]

for i in num:
    print(i)
````

````python
num = [1, 2, 3]

# Convert list (iterable) into an iterator object
iter_num = iter(num)

# next() fetches elements one by one from iterator
print(next(iter_num))  # 1st element: 1
print(next(iter_num))  # 2nd element: 2
print(next(iter_num))  # 3rd element: 3

# After last element, next() will raise StopIteration error
# because there are no more items left in the iterator
````

### Making own for loop

````python
# Custom loop function that works like a for loop internally

def own_loop(iterable):
    # Convert iterable into an iterator
    iterator = iter(iterable)

    # Infinite loop to keep fetching next element
    while True:
        try:
            # next() returns the next item from iterator
            print(next(iterator))

        # When no elements are left, StopIteration is raised
        except StopIteration:
            # Exit the loop when iteration is finished
            break


# Different types of iterables
a = [1, 2, 3]        # list
b = range(1, 11)     # range object
c = {1, 2, 3}        # set
d = (1, 2, 3)        # tuple
e = {0: 1, 1: 1}     # dictionary (iterates over keys)

# Calling custom loop function
own_loop(e)
````

````python
# A list is an iterable
num = [1, 2, 3]

# Convert list into an iterator
iter_obj = iter(num)

# Print memory address of iterator object
print(id(iter_obj), "Address of iterator 1")

# Calling iter() on an iterator returns the SAME iterator itself
iter_obj2 = iter(iter_obj)

# So both variables point to the same object
print(id(iter_obj2), "Address of iterator 2")
````

### Creating own range() function

````python
class own_range:
    def __init__(self, start, end):
        # Initialize starting and ending values of the range
        self.start = start
        self.end = end

    def __iter__(self):
        # Return an iterator object for this iterable (own_range)
        # This enables iteration using for loop or iter()
        return iterator_of_own_range(self)
````

````python
class iterator_of_own_range:
    def __init__(self, iterable_obj):
        # Store reference to the iterable object (own_range)
        # So we can access start and end values during iteration
        self.iterable = iterable_obj

    def __iter__(self):
        # An iterator must return itself
        # This allows it to be used in loops directly
        return self

    def __next__(self):
        # Stop iteration when start reaches or exceeds end
        if self.iterable.start >= self.iterable.end:
            raise StopIteration  # Signals end of iteration

        # Store current value to return
        current = self.iterable.start

        # Move start forward for next iteration
        self.iterable.start += 1

        # Return current value
        return current
````

### `Q1:` Create MyEnumerate class,
Create your own `MyEnumerate` class such that someone can use it instead of enumerate. It will need to return a `tuple` with each iteration, with the first element in the tuple being the `index` (starting with 0) and the second element being the `current element` from the underlying data structure. Trying to use `MyEnumerate` with a noniterable argument will result in an error.

```
for index, letter in MyEnumerate('abc'):
    print(f'{index} : {letter}')
```

Output:
```
0 : a
1 : b
2 : c
```

````python
class MyEnumerate:
    def __init__(self, data):
        # Store the input data (string, list, etc.)
        self.data = data

        # Maintain current index for iteration
        self.index = 0

    def __iter__(self):
        # Iterator returns itself
        return self

    def __next__(self):
        # Stop iteration when index goes beyond data length
        if self.index >= len(self.data):
            raise StopIteration

        # Create (index, value) pair like built-in enumerate()
        value = (self.index, self.data[self.index])

        # Move index forward for next call
        self.index += 1

        # Return current (index, value)
        return value


# Using custom enumerate in a for loop
for index, letter in MyEnumerate('abc'):
    print(f'{index} : {letter}')
````

### `Q2:` Iterate in circle
Define a class, `Circle`, that takes two arguments when defined: a sequence and a number. The idea is that the object will then return elements the defined number of times. If the number is greater than the number of elements, then the sequence  repeats as necessary. You can define an another class used as a helper (like I call `CircleIterator`).

```
c = Circle('abc', 5)
d = Circle('abc', 7)
print(list(c))
print(list(d))
```

Output
```
[a, b, c, a, b]
[a, b, c, a, b, c, a]
```

````python
# Custom circular iterator (loops over data repeatedly up to max_iters)

class Circle:

    def __init__(self, data, max_iters):
        # Store the data (string, list, etc.)
        self.data = data

        # Current position in iteration
        self.index = 0

        # Maximum number of iterations allowed
        self.max_iters = max_iters

    def __iter__(self):
        # Iterator returns itself
        return self

    def __next__(self):
        # Stop when maximum iterations are reached
        if self.index >= self.max_iters:
            raise StopIteration

        # Circular access using modulo
        value = self.data[self.index % len(self.data)]

        # Move to next position
        self.index += 1

        # Return current value
        return value


# Create Circle objects
c = Circle('abc', 5)
d = Circle('abc', 7)

# Convert iterator output to list
print(list(c))
print(list(d))
````
