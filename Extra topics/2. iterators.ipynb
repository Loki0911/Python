### `1.  Iteration`

Iterationn is a  general term for taking each item of something one after another.

Anytime you use a loop, explict or implict, to go over a group of items, that is iteration.

```python
L = [1,2,3]

for i in L:
    print(i)
```

### `2. Iterator`

Iterator is  an object that allows the  programmer to traverse through a sequence of data without having to store the entire data in memory.

```python
import sys
l = [i for i in range(1,100000)]

#for i in l:
        #print(i)

print(sys.getsizeof(l),'bytes')

x = i in range(1,100000)
print(sys.getsizeof(x),'bytes')
```

### ` 3.  Iterable`

Iterable is an object, which one can iterate over.

It generates an iterator object when passed to iter() method.

```python
L = [1,2,3]
type(L)

#L is  an iterable
type(iter(L))

#iter(L) -> iterator
```

## Points to remember:

- All iterator is also an iterable
- Not all iterables are iterators.

### Making own for loop

```python
def own_loop(iterable):
    iterator = iter(iterable)

    while True:
        try:
            print(next(iterator))

        except StopIteration:
            break

a = [1,2,3]
b = range(1,11)
c = {1,2,3}
d = (1,2,3)
e ={0:1,1:1}

own_loop(e)
```

### Creating own range() function

```python
class own_range:
    def __init__(self,start,end):
        self.start = start
        self.end = end

    def __iter__(self):
        return iterator_of_own_range(self)
```

```python
class iterator_of_own_range:
    def __init__(self,iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self

    def __next__(self):
        if self.iterable.start >= self.iterable.end :
            raise StopIteration

        current = self.iterable.start
        self.iterable.start += 1
        return current

        
```
