## Generators:

simple way to create iterators.

```python
#example

def gen_demo(): 
    yield "first statement"
    yield "second statement"
    yield "third statement"

gen = gen_demo()
# print(next(gen))
# print(next(gen))
# print(next(gen))
# print(next(gen))

for i in gen:
    print(i)
```

```python
 # Example 2

def square(num):
    for i in range(1,num+1):
        yield i**2

gen =  square(10)
print(next(gen))
print(next(gen))
print(next(gen))

for i in gen:
    print(i)
```

### creatign range() using decorators

```python
def my_range(start,end):

    for i in range(start,end):
        yield i

for i in my_range(15,30):
    print(i)
```

### Benifits of using generators:


### 1. Simple implementation

```python
class own_range:
    def __init__(self,start,end):
        self.start = start
        self.end = end

    def __iter__(self):
        return iterator_of_own_range(self)


class iterator_of_own_range:
    def __init__(self,iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self

    def __next__(self):
        if self.iterable.start >= self.iterable.end :
            raise StopIteration

        current = self.iterable.start
        self.iterable.start += 1
        return current

x = own_range(1,11)
type(x)
iter(x)
```

```python
#using generators

def my_range(start,end):

    for i in range(start,end):
        yield i

for i in my_range(15,30):
    print(i)
```

### 3. Representing infinite Streams

```python
def all_even():
    n=0
    while True:
        yield n
        n+=2

even_num_gen = all_even()
next(even_num_gen)
next(even_num_gen)
```

### 4. Chaning generators

```python
def fibonacci_numbers(num):
    x,y = 0,1

    for _ in range(num):
        x,y = y,x+y
        yield x

def square(nums):
    for num in nums:
        yield num**2

print(sum(square(fibonacci_numbers(10))))
```
