## Generators:

simple way to create iterators.

````python
# Example of generator function in Python

def gen_demo():
    # yield pauses function and returns value one at a time
    yield "first statement"
    yield "second statement"
    yield "third statement"


# Creating a generator object (not executing function immediately)
gen = gen_demo()

# next(gen) would manually fetch values one by one
# print(next(gen))
# print(next(gen))
# print(next(gen))
# print(next(gen))  # would raise StopIteration

# for loop automatically calls next() until StopIteration
for i in gen:
    print(i)
````

````python
# Example of generator that produces square numbers

def square(num):
    # Loop from 1 to num
    for i in range(1, num + 1):
        # yield returns square of i one by one
        # pauses function and saves state after each yield
        yield i ** 2


# Create generator object (function does not execute fully yet)
gen = square(10)

# Manually fetching first 3 values using next()
print(next(gen))  # 1^2 = 1
print(next(gen))  # 2^2 = 4
print(next(gen))  # 3^2 = 9

# Remaining values are automatically fetched by for loop
for i in gen:
    print(i)
````

### creatign range() using decorators

````python
# Custom generator function similar to built-in range()

def my_range(start, end):
    # Loop from start to end-1
    for i in range(start, end):
        # yield returns values one by one (lazy execution)
        yield i


# Using generator in a for loop
for i in my_range(15, 30):
    print(i)
````

### Practical example

````python
import os
import cv2

# Generator function to read images from a folder one by one
def image_data_reader(folder_path):

    # os.listdir() gives list of all files in the folder
    for file in os.listdir(folder_path):

        # Build full path of each image file
        # cv2.imread() reads image and converts it into numpy array
        f_array = cv2.imread(os.path.join(folder_path, file))

        # yield returns one image at a time (lazy loading)
        yield f_array
````

### 1. Simple implementation

````python
class own_range:
    def __init__(self,start,end):
        self.start = start
        self.end = end

    def __iter__(self):
        return iterator_of_own_range(self)


class iterator_of_own_range:
    def __init__(self,iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self

    def __next__(self):
        if self.iterable.start >= self.iterable.end :
            raise StopIteration

        current = self.iterable.start
        self.iterable.start += 1
        return current

x = own_range(1,11)
type(x)
iter(x)
````

### 3. Representing infinite Streams

````python
def all_even():
    # Start from 0 (first even number)
    n = 0

    # Infinite loop to generate even numbers forever
    while True:
        # yield current even number and pause function here
        yield n

        # move to next even number
        n += 2


# Create generator object
even_num_gen = all_even()

# Fetch even numbers one by one using next()
next(even_num_gen)  # 0
next(even_num_gen)  # 2
````

### 4. Chaning generators

````python
def fibonacci_numbers(num):
    # First two Fibonacci numbers
    x, y = 0, 1

    # Generate Fibonacci sequence up to 'num' terms
    for _ in range(num):

        # Update Fibonacci values
        x, y = y, x + y

        # Yield current Fibonacci number
        yield x


def square(nums):
    # Take a sequence and yield square of each number
    for num in nums:
        yield num ** 2


# Step 1: Generate first 10 Fibonacci numbers
# Step 2: Square each Fibonacci number using generator chaining
# Step 3: Sum all squared values
print(sum(square(fibonacci_numbers(10))))
````

### `Q1:` Generator time elapsed
Write a generator function whose argument must be iterable. With each iteration, the generator will return a two-element tuple. The first element in the tuple will be an integer indicating how many seconds have passed since the previous iteration. The tuple’s second element will be the next item from the passed argument.

Note that the timing should be relative to the previous iteration, not when the
generator was first created or invoked. Thus the timing number in the first iteration
will be 0

```
for t in elapsed_since('abcd'):
    print(t)
    time.sleep(2)
```

Output:
```
(0.0, 'a')
(2.005651817999933, 'b')
(2.0023095009998997, 'c')
(2.001949742000079, 'd')
```
Note: Your output may differ because of diffrent system has different processing configuration.

````python
import time

# Generator that yields (time gap, item) for each element in iterable
def elapsed_since(iterable):

    # Store initial timestamp
    last_time = time.perf_counter()

    # Iterate over each item in the iterable
    for item in iterable:

        # Get current time before processing item
        current_time = time.perf_counter()

        # Calculate time difference from previous item
        delta = current_time - last_time

        # Update last_time for next iteration
        last_time = current_time

        # Yield time difference along with current item
        yield (delta, item)


# Loop through generator output
for t in elapsed_since('abcd'):
    print(t)

    # Pause execution for 2 seconds between iterations
    time.sleep(2)
````
