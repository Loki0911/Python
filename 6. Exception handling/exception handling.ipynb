### ` Catching specific exception`

```python


try:
    # object of file created, raises FileNotFoundError if file does not exist
    f = open('sample.txt', 'r')

    m = 5

    print(f.read())  # reads file content
    print(m)         # prints variable m
    print(5/2)       # no ZeroDivisionError here since divisor is 2 not 0

    l = [1, 2, 3]
    l[100]           # raises IndexError since list only has 3 elements

except FileNotFoundError:
    # object of FileNotFoundError caught when file is missing
    print('File not found')

except NameError:
    # object of NameError caught when undeclared variable is used
    print('Variable not declared')

except ZeroDivisionError:
    # object of ZeroDivisionError caught when dividing by zero
    print("Can't divide by zero")

except Exception as e:
    # object of any other Exception caught here as 'e'
    # IndexError from l[100] will be caught here since no specific handler for it
    print(e)
```

### `else block`

```python
# else block

try:
    # object of file created, raises FileNotFoundError if file does not exist
    f = open('sample.txt', 'r')

except FileNotFoundError:
    # object of FileNotFoundError caught here when file is missing
    print("File not found")

except Exception as e:
    # object of any other Exception caught here as 'e' for unexpected errors
    print("There is something wrong")

else:
    # else block runs only if no exception was raised in try block
    # file opened successfully, so we read and print its content here
    print(f.read())
```

### `finally block`

```python
# Finally block

try:
    # object of file created, raises FileNotFoundError if file does not exist
    f = open('sample.txt', 'r')

except FileNotFoundError:
    # object of FileNotFoundError caught here when file is missing
    print("File not found")

except Exception as e:
    # object of any other Exception caught here as 'e' for unexpected errors
    print("There is something wrong")

else:
    # runs only if no exception was raised, file opened successfully
    print(f.read())

finally:
    # always runs regardless of exception or not, good place to release resources
    print("This will be surely printed")
```

```python
# example

class Bank:

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):

        # object of built-in Exception class created and raised when amount is negative
        if amount < 0:
            raise Exception('amount cannot be -ve')

        # object of built-in Exception class created and raised when balance is insufficient
        if self.balance < amount:
            raise Exception("Not that much money in your account")

        else:
            self.balance = self.balance - amount


obj = Bank(10000)

try:
    # withdraw called with -5000, triggers object creation of built-in Exception class
    obj.withdraw(-5000)

except Exception as e:
    # object of built-in Exception class caught here as 'e'
    print(e)  # prints the message stored in the Exception object

else:
    print(obj.balance)
```

### `Creating custom exception class`

```python
# creating custom exception class by inheriting from built-in Exception class
class MyException(Exception):

    def __init__(self, message):
        print(message)


class Bank:

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):

        # object of MyException created and raised when amount is negative
        if amount < 0:
            raise MyException('amount cannot be -ve')

        # object of MyException created and raised when balance is insufficient
        if self.balance < amount:
            raise MyException("Not that much money in your account")
        
        else:
            self.balance = self.balance - amount


obj = Bank(10000)

try:
    # withdraw called with -5000, triggers object creation of MyException
    obj.withdraw(-5000)

except MyException as e:
    # object of MyException caught here as 'e'
    pass

else:
    print(obj.balance)
```

```python
# creating custom exception class by inheriting from built-in Exception class
class SecurityError(Exception):

    def __init__(self, message):
        print(message)  # object of SecurityError created, message printed

    def logout(self):
        print("You have been logged out")  # custom method only possible in custom exception class


class Google:

    def __init__(self, name, email, password, device):
        self.name = name
        self.email = email
        self.password = password
        self.device = device

    def login(self, email, password, device):

        # object of SecurityError created and raised when device does not match
        if device != self.device:
            raise SecurityError('Unusual login attempt')

        if email == self.email and password == self.password:
            print("Login successful")
        else:
            print("Login Error")


obj = Google('Lokesh', 'lokesh@gmail.com', '1234', 'android')

try:
    # login attempted from 'Windows', triggers object creation of SecurityError
    obj.login('lokesh@gmail.com', '1234', 'Windows')

except SecurityError as e:
    # object of SecurityError caught here as 'e', logout method called on it
    e.logout()

else:
    print(obj.name)

finally:
    # always runs regardless of exception
    print('Database connection closed')
```

### Why we need custom exception classes?
Custom exception classes let you add specific behavior and meaning beyond what built-in exceptions offer, as seen with SecurityError having its own logout() method. When a wrong device triggers the exception, the except block handles it specifically, else is skipped, and finally always runs regardless. The key fix needed is using super().__init__(message) instead of just print(message) so the exception message is properly stored.
