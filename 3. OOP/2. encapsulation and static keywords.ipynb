### `Q.` Write OOP classes to handle the following cases
 1. A user can create and view 2D coordinates
 2. A user can find out the distance between 2 coordinates
 3. A user can fins the distance of a coordinate from origin
 4. A user can check if a point lies on a given line
 5. A user can find the distance between a given 2D point and a given line

```python
class Point:

    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y
    
    def __str__(self):
        return '({},{})'.format(self.x_cod,self.y_cod)

    def euclidean_distance(self,other):
        return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5 
    
    def distance_from_origin(self):
        return self.euclidean_distance(Point(0,0))


class Line:

    def __init__(self,A,B,C):
        self.A = A
        self.B = B
        self.C = C
    
    def __str__(self):
        return '{}x + {}y + {} = 0'.format(self.A,self.B,self.C)

    def point_on_line(line,point):
        if line.A*point.x_cod + line.B * point.y_cod + line.C == 0:
            return "point lies on the line"
        else:
            return "point does not lie on the line"
    
    def shortest_distance(line,point):
        return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5
    
    def interset_line(self,other):
         # Solve using Cramer's rule:
        # A1x + B1y = -C1
        # A2x + B2y = -C2
        determinant = self.A * other.B - other.A * self.B
    
        if determinant == 0:
            return "Lines are parallel, no intersection"
        
        x = (-self.C * other.B + other.C * self.B) / determinant
        y = (-self.A * other.C + other.A * self.C) / determinant
        
        return Point(x, y)
```

```python
p1 = Point(2, 3)        # point at (2,3)

l1 = Line(1, 1, -5)     # line: x + y - 5 = 0

print(l1.point_on_line(p1))       # 1(2) + 1(3) - 5 = 0 → lies on line
print(l1.shortest_distance(p1))   # point on line → distance = 0.0
```

## `1. Concept of accessing attributes and methods of the class`

```python
class Person:

    def __init__(self, name, country):
        self.name = name
        self.country = country

    def greet(self):
        # greet in local style based on country
        if self.country == 'Nepal':
            print('Namaste ', self.name)
        else:
            print('Hello ', self.name)
```

` ## 4. Mutablity of an object`

```python
class Person:

    def __init__(self,name,gender):
        self.name=name
        self.gender = gender
    

#outside the class
def greet(person):
    person.name = "Mahendra"
    return person

p=Person('Lokesh','male')
print(id(p))
p1 = greet(p)
print(id(p1))
```

## ` 7. Static Keywords `

```python
# need of static variables
class Atm:
    __counter = 1
    def __init__(self):
        self.pin = ''
        self.__balance = 0
        self.cid = Atm.__counter
        Atm.__counter += 1

    #utility methods
    @staticmethod
    def get_counter():
        return Atm.__counter
    
    def get_balance(self):
        return self.__balance
    
    def set_balance(self, new_value):
        if type(new_value) == int :
        
            self.__balance = new_value
        else:
            print("no other type supported")

    def menu(self):
        while True:                          # ← loop instead of recursion
            user_input = input("""
        Hi how can I help you?
        1. Press 1 to create Pin
        2. Press 2 to change Pin
        3. Press 3 to check balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)
            if user_input == '1':
                self.create_pin()
            elif user_input == '2':
                self.change_pin()
            elif user_input == '3':
                self.check_balance()
            elif user_input == '4':
                self.withdraw_cash()
            else:
                print("Goodbye!")
                break                        # ← clean exit

    def create_pin(self):
        self.pin = input("Enter your pin: ")
        self.__balance = int(input('Enter the balance: '))
        print('Pin created successfully')
        # ← no self.menu() call here

    def change_pin(self):
        old_pin = input("Enter your current pin: ")
        if old_pin == self.pin:
            self.pin = input("Enter new pin: ")
            print("Pin changed successfully")
        else:
            print("Incorrect existing pin")
        # ← no self.menu() call here

    def check_balance(self):
        user_pin = input("Enter the pin: ")
        if user_pin == self.pin:
            print("Your current balance is {}".format(self.__balance))
        else:
            print("Incorrect pin")

    def withdraw_cash(self):
        user_pin = input("Enter the pin: ")
        if user_pin == self.pin:
            cash = int(input('Enter the amount you want to withdraw: '))
            if cash <= self.__balance:
                self.__balance -= cash
                print("{} withdrawn successfully".format(cash))
            else:
                print("Insufficient balance")
        else:
            print("Incorrect pin")
```
