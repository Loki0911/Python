```python
import json    
```

```python
#function to load file
def load_data(filename):
    with open(filename,'r') as f:
        return json.load(f)

#function for sugestion of friends
def people_you_may_know( user_id, data):
    user_friends={}
    for user in data['users']:
        user_friends[user['id']]=set(user['friends'])        #mapping user id with the friends

    if user_id not in user_friends:             #if  inputed user id doesnot in data
        return []                              #return null
        
    direct_friends=user_friends[user_id]
    suggestions={}

    #loop through direct friends of user
    for friend in direct_friends:

        #loop through each friend of direct friend of user
        for mutual in user_friends[friend]:

            #dont innclude user himself and the direct friends of user
            if mutual!=user_id and mutual not in direct_friends:
                #count people user may know
                suggestions[mutual]=suggestions.get(mutual,0)+1

     #sorting in descending order
    sorted_suggestions=sorted(suggestions.items(),key=lambda x:x[1], reverse=True)

    #returning user id inputed user may know
    return [user_id for user_id, _ in sorted_suggestions]
data=load_data("massive_data.json")
user_id=int(input("Enter user id to know people he may know (1 to 30) : "))
recommendation = people_you_may_know( user_id, data)
print(f" people user-{user_id} may know: {recommendation}")

    
```
