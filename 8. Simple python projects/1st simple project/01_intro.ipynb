```python
import json
```

```python
#function to load users data
def load_data(filename):
    with open(filename,'r') as f:
        data=json.load(f)
    return data
```

```python
data=load_data("data.json")
```

```python
#to display users and their connection
def display_users(data):
    print("users And their connections:\n")
    for user in data['users']:
        print(f"ID:{user['id']}-{user['name']} is friend with:{user['friends']} and liked pages are {user['liked_pages']}")
    print("\n\npages Information:\n")
    for pages in data['pages']:
        print(f"{pages['id']}:{pages['name']}")
    
        
```

```python
display_users(data)
```
