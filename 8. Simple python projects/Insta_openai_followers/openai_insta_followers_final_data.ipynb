```python
with open("finaldata.txt",encoding="utf-8") as f:
    data=f.read()
    
```

```python
chunks=data.split("\n\n")
chunks=[c for c in chunks if len(c)>5]
```

```python
def pharse_chunk(chunk):
    chunk=chunk.strip()
    lines=chunk.split("\n")
    user_name=lines[0]
    posts=int(lines[1].split("post")[0].replace(",",""))
    no_of_followers=float(lines[2].split("follower")[0].replace(",","").replace("K","").replace("M",""))
    if ("K" in lines[2]):
        no_of_followers=int(no_of_followers*1000)
    elif ("M" in lines[2]):
        no_of_followers=int(no_of_followers*1000000)
    else:
        no_of_followers=int(no_of_followers)
    following=float(lines[3].split("following")[0].replace(",","").replace("K","").replace("M",""))
    if ("K" in lines[3]):
        following=int(following*1000)
    elif ("M" in lines[3]):
        following=int(following*1000000)
    else:
        following=int(following)
    name=lines[4]
    if(len(lines)>5):
        type_of_page=lines[5]
        bio="\n".join(lines[6:])
    else:
        type_of_page="Not mentioned"
        bio=""
   # print(user_name,posts,no_of_followers,following,name,type_of_page,bio,sep="\n")
    return({"User Name":user_name,
            "Posts":posts,
            "Number of followers":no_of_followers,
            "Following":following,
            "Name":name,
            "Type of Page": type_of_page,
            "Bio":bio})
```

```python
all_chunks=[]
for chunk in chunks:
    pharsed_chunk=pharse_chunk(chunk)
    all_chunks.append(pharsed_chunk)
```

```python
import json
s=json.dumps(all_chunks,indent=4)
with open("data.json","w") as f:
    f.write(s)

```

```python
#Who has maximum posts
maxm=0
for chunk in all_chunks:
    if (maxm<chunk['Posts']):
        maxm=chunk['Posts']
        chunk_with_maxm_posts=chunk
print(f"{chunk_with_maxm_posts['User Name']} with {chunk_with_maxm_posts['Posts']} posts")
```

```python
#Who has Maximum Followers
max_followers=0
for chunk in all_chunks:
    if(max_followers<chunk['Number of followers']):
        max_followers=chunk['Number of followers']
        chunk_with_maxm_followers=chunk
print(f"{chunk_with_maxm_followers['User Name']} with {chunk_with_maxm_followers['Number of followers']} followers")
```

```python
max_following=0
for chunk in all_chunks:
    if(max_following<chunk['Following']):
        max_following=chunk['Following']
        chunk_with_maxm_following=chunk
print(f"{chunk_with_maxm_following['User Name']} with  {chunk_with_maxm_following['Following']} following")
```

```python
#How many categories are there
categories=[]
for chunk in all_chunks:
    categories.append(chunk["Type of Page"])
categories=list(set(categories))
print(f" There are {len(categories)} categories")
```
